```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os
try:
    os.mkdir('augmented_images')
except:
    pass
print(f'증강전 : {os.listdir("augmented_images")}')
img_gen = ImageDataGenerator(
    shear_range=0.2,
    zoom_range=0.2,
    width_shift_range=0.2,
    height_shift_range=0.2,
    rescale=1./255,
    rotation_range=40,
    vertical_flip=True,
    horizontal_flip=True
)
img_generator = img_gen.flow_from_directory(
    './images',  # 원본 이미지가 있는 디렉토리
    target_size=(150, 150), # 증강된 이미지 크기
    batch_size=1, # 한 번에 처리할 이미지 수 (저장 시 1로 설정하면 편리)
    class_mode='categorical', # 분류 모델을 위한 모드
    save_to_dir='augmented_images',  # ⭐ 새로 추가된 코드: 증강된 이미지를 저장할 디렉토리
    save_prefix='aug', # 저장될 파일명에 붙을 접두사
    save_format='png' # 저장될 이미지 파일 형식
)
for i in range(1000):
    next(img_generator) 
print(f'증강후 : {os.listdir("augmented_images")}')
```
